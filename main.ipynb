```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('House Price Prediction Dataset.csv')
df
```

```python
weight = 0.0
bias = 0.0
x_data = df['Area'].values / df['Area'].max()
y_data = df['Price'].values / df['Price'].max()
```

```python
def predict(x_data):
    y_predict = weight * x_data + bias
    return y_predict
```

```python
def error(y_data, y_predict):
    return  y_predict - y_data
```

```python
def loss(error):
    loss = np.mean(error ** 2)
    return loss 
```

```python
def optimize(x_data, y_data):
    global weight, bias
    epochs = range(1000)
    learning_rate = 0.01
    n = len(x_data)
    for epoch in epochs:

        pred = predict(x_data)
        err = error(y_data, pred)

        d_weight = (2/n) * np.sum(x_data * err)
        d_bias = (2/n) * np.sum(err) 

        weight -= learning_rate * d_weight
        bias -= learning_rate * d_bias

        losses = loss(err)

        print(f"loss: {losses}, epoch: {epoch}")
optimize(x_data, y_data)
```
